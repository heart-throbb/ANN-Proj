```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder
import pickle
from tensorflow.keras.models import load_model
```

```python
# Load all th trained model,scalar pickel, onehot
model=load_model('model.h5')

# load encoder and scalar
with open("gender_encoder.pkl",'rb') as file:
    gender_encoder=pickle.load(file)

with open("geography_encoder.pkl",'rb') as file:
    geography_encoder=pickle.load(file)
    
with open("scaler.pkl",'rb') as file:
    scalar=pickle.load(file)
```

```python
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'Exited': 0
}
```

```python
geo_encoded=geography_encoder.transform([[input_data['Geography']]]).toarray()
geo_encoded_df=pd.DataFrame(geo_encoded,columns=geography_encoder.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
input_df=pd.DataFrame([input_data])
input_df
```

```python
# Encode the categorical cols
input_df['Gender']=gender_encoder.transform([input_df['Gender']])
```

```python
input_df = pd.concat([input_df.drop("Geography", axis=1),geo_encoded_df], axis=1)
```

```python
input_df
```

```python
expected_columns = [
    "CreditScore",
    "Gender",
    "Age",
    "Tenure",
    "Balance",
    "NumOfProducts",
    "HasCrCard",
    "IsActiveMember",
    "Exited",
    "Geography_France",
    "Geography_Germany",
    "Geography_Spain"
]

input_df = input_df[expected_columns]


print("\nProcessed Input:")
print(input_df)
```

```python
scalaer_input=scalar.transform(input_df)
scalaer_input
```

```python
prediction=model.predict(scalaer_input)
prediction
```

```python
predicted_salary = prediction[0][0]
predicted_salary
```

```python
print("ESTIMATED SALARY PREDICTION")
print(f"Predicted Estimated Salary: {predicted_salary:.2f}")
```
